## Шаг 2. Инжиниринг признаков

### 2.1. Группы по срокам финансирования

Разделим все компании на три группы:

* Единичное финансирование — был всего один раунд финансирования.

* Срок финансирования до года — между первым и последним раундом финансирования прошло не более года.

* Срок финансирования более года.

Визуализируем соотношение этих групп, создав два графика:

* По количеству компаний: Покажите, какой процент от общего числа компаний относится к каждой из трёх групп.
* По объёму инвестиций: Отобразите, какую долю от общего объёма привлечённых средств получила каждая группа.

Визуализируем соотношения групп по количеству компаний

```python
# Вычислим разницу лет между первым и полседним раундом финансирования 
df_inv['diff_funding_at'] = (df_inv['last_funding_at'] - df_inv['first_funding_at']) / np.timedelta64(1, 'Y')

# Задаем условия для деления на категории
conditions = [
    df_inv['diff_funding_at'] == 0,
    df_inv['diff_funding_at'] < 1,
    df_inv['diff_funding_at'] >= 1
]

# Задаем названия категорий для каждого условия
categories = ['единичное финансирование', 'срок до года', 'срок более года']

# Применяем условия
df_inv['funding_category'] = np.select(conditions, categories, default=np.nan) 
```

Оставим в столбце `market` только массовые сегменты. Для остальных сегментов заменим значения на заглушки — `niche` для нишевых и `mid` для средних.
Дальнейшие исследования будем выполнять с учётом этой замены.

```python
# Создадим функцию для замены значений  
def categorize_market(market_name):
    # Если значение пропущено, оставляем его без изменений
    if pd.isna(market_name):
        return market_name
        
    count = market_counts[market_name]
    
    if count > 120:
        return market_name       # Оставляем массовые сегменты как есть
    elif 35 <= count <= 120:
        return 'mid'              # Заглушка для средних
    else:
        return 'niche'            # Заглушка для нишевых

# 3. Применяем функцию к столбцу market
df_inv['market'] = df_inv['market'].apply(categorize_market)
```

Определим компании с аномальным объёмом общего финансирования — используем метод IQR отдельно по каждому сегменту. Напомним, что все нишевые сегменты должны быть объединены в одну группу, а средние — в другую.

Определим сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, и выведем топ таких сегментов.

```python
# Найдем компании с аномальными выбросами в нишевом сегменте
df_inv_niche = df_inv[df_inv['market']=='niche'].copy()

Q1 = df_inv_niche['funding_total_usd'].quantile(0.25)
Q3 = df_inv_niche['funding_total_usd'].quantile(0.75)
QM = df_inv_niche['funding_total_usd'].quantile(0.5)
IQR = Q3 - Q1

# 2. Определяем границы для поиска аномалий
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 3. Фильтруем только аномально высокие значения (верхние выбросы)
anomalies = df_inv_niche[df_inv_niche['funding_total_usd'] > upper_bound]

# 4. Выводим названия компаний и их финансирование
result_niche = anomalies[['name', 'funding_total_usd']].sort_values(by='funding_total_usd', ascending=False)
print(result_niche)
```

### 4.1 Динамика предоставления финансирования по годам

На основе получившейся таблицы постройте графики, отражающие:
* динамику типичного размера средств, которые стартапы получали в рамках одного раунда финансирования;

* динамику общего количества раундов за каждый год, то есть насколько активно происходили инвестиции на рынке (чем больше раундов, тем выше активность).

Когда будете строить графики в этом задании и следующих, используйте данные только по тем компаниям, которые остались в датасете после предыдущих фильтраций.

На основе полученных данных ответьте на вопросы:
* В каком году типичный размер средств, собранных в рамках одного раунда, был максимальным?

* Какая тенденция наблюдалась в 2014 году по количеству раундов и средств, выделяемых в рамках каждого раунда?

### 4.3 Годовая динамика доли возвращённых средств по типам финансирования

Заказчик хочет знать, какая часть вложенных или выданных денег со временем возвращается обратно инвесторам или финансистам. Ваша цель — для каждого года и каждого вида финансирования рассчитать нормированные значения возврата средств: то есть какую долю возвращённые средства составляют от предоставленных. При этом слишком большие аномальные значения, то есть неадекватные выбросы, нужно заменить на пропуски.

Совет: когда будете делить сумму возвращённых средств на суммарный объём привлечённого финансирования по конкретному году, добавьте к знаменателю небольшое число, например `1e-60`. Это поможет избежать деления на ноль.

```python
df_inv['mid_funding_at'] = pd.to_datetime(df_inv['mid_funding_at'])

df_filtered = df_inv[df_inv['mid_funding_at'].dt.year >= 2000]

funding_cols = ['seed', 'venture', 'equity_crowdfunding', 'undisclosed', 
                'convertible_note', 'debt_financing', 'angel', 'grant', 
                'private_equity', 'post_ipo_equity', 'post_ipo_debt', 
                'secondary_market', 'product_crowdfunding']

funding_by_year = df_filtered.groupby(df_filtered['mid_funding_at'].dt.year)[funding_cols].sum().rename_axis('year')

funding_by_year
```

```python
# умножим значения в df_ret на 1000000 чтобы вычисления были с цифрами одного порядка (df_ret все значения указаны в млн)
df_ret = df_ret *1000000
```

```python
# Соединим 2 датасета
df_merged = pd.merge(funding_by_year, df_ret, on="year", how="inner")

#Рассчитаем доли
df_merged["seed_ratio"] = df_merged["seed_y"] / (df_merged["seed_x"]+ 1e-60)
df_merged["venture_ratio"] = df_merged["venture_y"] / (df_merged["venture_x"] + 1e-60)
df_merged["equity_crowdfunding_ratio"] = df_merged["equity_crowdfunding_y"] / (df_merged["equity_crowdfunding_x"] + 1e-60)
df_merged["convertible_note_ratio"] = df_merged["convertible_note_y"] / (df_merged["convertible_note_x"] + 1e-60)
df_merged["debt_financing_ratio"] = df_merged["debt_financing_y"] / (df_merged["debt_financing_x"] + 1e-60)
df_merged["angel_ratio"] = df_merged["angel_y"] / (df_merged["angel_x"] + 1e-60)
df_merged["grant_ratio"] = df_merged["grant_y"] / (df_merged["grant_x"] + 1e-60)
df_merged["private_equity_ratio"] = df_merged["private_equity_y"] / (df_merged["private_equity_x"] + 1e-60)
df_merged["post_ipo_equity_ratio"] = df_merged["post_ipo_equity_y"] / (df_merged["post_ipo_equity_x"] + 1e-60)
df_merged["post_ipo_debt_ratio"] = df_merged["post_ipo_debt_y"] / (df_merged["post_ipo_debt_x"] + 1e-60)

# Заменим аномальные значения на пропуски
df_merged = df_merged.replace([np.inf, -np.inf], np.nan)

df_merged
```

Постройте график, на котором отобразите нормированные значения возврата средств для типов финансирования `venture`, `debt_financing`, `private_equity`, `seed` и `angel`.

Сделайте вывод о том, в каких типах финансирования наблюдается наиболее устойчивый рост показателя.

```python
# Список колонок, которые нужно оставить
target_columns = ['venture_ratio', 'debt_financing_ratio', 'private_equity_ratio', 'seed_ratio', 'angel_ratio']

# Перезапись DataFrame только с этими колонками
df_merged = df_merged[target_columns]

# Построение графика
df_merged.plot(kind="line", marker="o", figsize=(20, 10))


plt.title("Динамика нормированных значений возврата средств по годам")
plt.xlabel("Год")
plt.ylabel("Значение")
plt.grid(True) 

plt.legend(
    title="Тип финасирования",
    title_fontsize=13,
    fontsize=11,
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=True,
)

plt.show()
```

Доля возвращенных средств только для  `private_equity` в 2014 незначительно снизилась. Все остальные показывают рост по сравнению с 2013 годом. Самый активный рост у `angel`
